```python
df_sales = (spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/batchoct26/trainingschema/trainingvolume/sales.csv")
)

df_sales.display()
```

```python
df_sales.write.saveAsTable("batchoct26.trainingschema.sales_table")
```

```python
%sql
select * from sales_table
```

```python
df_sales.write.format("delta").saveAsTable("batchoct26.trainingschema.sales_delta_table")
```

```python
%sql
SELECT product, SUM(amount) AS total_amount
FROM batchoct26.trainingschema.sales_delta_table
GROUP BY product
ORDER BY total_amount DESC
```

```python
from pyspark.sql.functions import sum as _sum

(df_sales.groupBy("product")
    .agg(_sum("amount").alias("total_amount"))
    .orderBy("total_amount", ascending=False)
    .display())
```
